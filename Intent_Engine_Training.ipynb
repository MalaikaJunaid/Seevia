## Load and Prepare Data

### Subtask:
Load the `multilingual_intent_dataset.csv` file, convert it into a Hugging Face `Dataset` object, split it into training and testing sets, and encode the intent labels into numerical IDs.


**Reasoning**:
Load the `multilingual_intent_dataset.csv` file into a Pandas DataFrame, then extract unique intent labels, create a mapping to numerical IDs, and apply this mapping to create a new 'labels' column in the DataFrame.



```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Load the multilingual_intent_dataset.csv file into a Pandas DataFrame
df = pd.read_csv('multilingual_intent_dataset.csv')

# 2. Create a list of unique intent labels
unique_intents = df['intent'].unique().tolist()

# 3. Create a dictionary that maps each unique intent label to a numerical ID
label_encoder = LabelEncoder()
df['labels'] = label_encoder.fit_transform(df['intent'])

label_to_id = {label: idx for idx, label in enumerate(label_encoder.classes_)}
id_to_label = {idx: label for idx, label in enumerate(label_encoder.classes_)}

print("DataFrame head after adding 'labels' column:")
print(df.head())
print("\nUnique Intents:", unique_intents)
print("Label to ID mapping:", label_to_id)
```

**Reasoning**:
The next step is to convert the Pandas DataFrame into a Hugging Face Dataset object and then split it into training and testing sets, ensuring stratification by labels as specified in the instructions.



```python
from datasets import Dataset, DatasetDict

# 5. Convert the Pandas DataFrame df into a Hugging Face Dataset object.
hf_dataset = Dataset.from_pandas(df, preserve_index=False)

# 6. Split the Dataset object into training and testing sets (80% train, 20% test).
#    Ensuring stratification by labels.
train_test_split_dataset = hf_dataset.train_test_split(test_size=0.2, stratify_by_column='labels')

# Assign to train_dataset and test_dataset for clarity
train_dataset = train_test_split_dataset['train']
test_dataset = train_test_split_dataset['test']

print(f"Total dataset size: {len(hf_dataset)}")
print(f"Training dataset size: {len(train_dataset)}")
print(f"Testing dataset size: {len(test_dataset)}")
print("Train dataset head:")
print(train_dataset[0])
```

**Reasoning**:
The error `ValueError: Stratifying by column is only supported for ClassLabel column, and column labels is Value.` indicates that the 'labels' column needs to be explicitly defined as a `ClassLabel` feature within the Hugging Face Dataset before stratification can be applied. I will cast the 'labels' column to a ClassLabel type with the correct number of classes.



```python
from datasets import Dataset, DatasetDict, ClassLabel, Features, Value
import pandas as pd

# Ensure df is available from previous steps
# If this were a new session, you'd reload df here
# df = pd.read_csv('multilingual_intent_dataset.csv')
# label_encoder = LabelEncoder()
# df['labels'] = label_encoder.fit_transform(df['intent'])
# id_to_label = {idx: label for idx, label in enumerate(label_encoder.classes_)}

# 5. Convert the Pandas DataFrame df into a Hugging Face Dataset object.
hf_dataset = Dataset.from_pandas(df, preserve_index=False)

# Define the features with 'labels' as a ClassLabel
features = hf_dataset.features.copy()
features['labels'] = ClassLabel(num_classes=len(id_to_label), names=list(id_to_label.values()))

# Cast the dataset to include the ClassLabel feature for 'labels'
hf_dataset = hf_dataset.cast(features)

# 6. Split the Dataset object into training and testing sets (80% train, 20% test).
#    Ensuring stratification by labels.
train_test_split_dataset = hf_dataset.train_test_split(test_size=0.2, stratify_by_column='labels', seed=42)

# Assign to train_dataset and test_dataset for clarity
train_dataset = train_test_split_dataset['train']
test_dataset = train_test_split_dataset['test']

print(f"Total dataset size: {len(hf_dataset)}")
print(f"Training dataset size: {len(train_dataset)}")
print(f"Testing dataset size: {len(test_dataset)}")
print("Train dataset head:")
print(train_dataset[0])
print("Test dataset head:")
print(test_dataset[0])
```

## Initialize Tokenizer

### Subtask:
Load the `DistilBertTokenizerFast` for `distilbert-base-multilingual-cased` to process the text data.


#### Reasoning:
To prepare our text data for the DistilBERT model, we need to use the tokenizer specifically designed for it. `DistilBertTokenizerFast` from the `transformers` library is efficient and correctly handles the tokenization rules (e.g., word segmentation, special tokens, padding) for the `distilbert-base-multilingual-cased` model, which is crucial for model compatibility and performance. This step will load the pre-trained tokenizer, making it ready to process our dataset's 'text' column.

**Reasoning**:
Following the instructions, I will now import the necessary tokenizer and initialize it with the specified pre-trained model. This code block will load the `distilbert-base-multilingual-cased` tokenizer.



```python
from transformers import DistilBertTokenizerFast

# Initialize the tokenizer by loading the pre-trained distilbert-base-multilingual-cased tokenizer
tokenizer = DistilBertTokenizerFast.from_pretrained('distilbert-base-multilingual-cased')

print("Tokenizer loaded successfully.")
print(f"Tokenizer vocabulary size: {len(tokenizer)}")
```

**Reasoning**:
The previous step successfully loaded the tokenizer. Now, I will define a tokenization function to apply the `tokenizer` to the 'text' column of both the training and testing datasets. This function will convert the text into numerical IDs and prepare it for model input.



```python
def tokenize_function(examples):
    return tokenizer(examples['text'], truncation=True, padding='max_length')

# Apply the tokenization function to both train and test datasets
tokenized_train_dataset = train_dataset.map(tokenize_function, batched=True)
tokenized_test_dataset = test_dataset.map(tokenize_function, batched=True)

print("Tokenization complete for training and testing datasets.")
print("First tokenized training example:", tokenized_train_dataset[0])
```

## Configure Model and Training Arguments

### Subtask:
Load the `DistilBertForSequenceClassification` model, specifying the number of unique intent labels, and define `TrainingArguments` for fine-tuning parameters such as learning rate, batch size, and number of epochs.


**Reasoning**:
The subtask requires loading the DistilBertForSequenceClassification model and defining TrainingArguments. I will import the necessary classes, load the model with the correct number of labels, and then instantiate TrainingArguments with the specified parameters.

```python
from transformers import DistilBertForSequenceClassification, TrainingArguments

# 1. Load a pre-trained DistilBertForSequenceClassification model
num_labels = len(id_to_label)
model = DistilBertForSequenceClassification.from_pretrained('distilbert-base-multilingual-cased', num_labels=num_labels)

# 2. Define TrainingArguments
training_args = TrainingArguments(
    output_dir='./results',          # output directory
    num_train_epochs=3,              # total number of training epochs
    per_device_train_batch_size=16,  # batch size per device during training
    per_device_eval_batch_size=64,   # batch size for evaluation
    warmup_steps=500,                # number of warmup steps for learning rate scheduler
    weight_decay=0.01,               # strength of weight decay
    logging_dir='./logs',            # directory for storing logs
    logging_steps=10                # log every x updates steps
    # evaluation_strategy='epoch'    # Removed as it caused TypeError
)

print("Model configuration:")
print(model.config)
print("\nTraining arguments:")
print(training_args)
```

**Reasoning**:
The previous step successfully initialized the model and training arguments. The next logical step in fine-tuning a Hugging Face model is to set up the `Trainer` object, which orchestrates the training process. This involves defining the model, training arguments, datasets, and a data collator, as well as a metric for evaluation during training.



```python
from transformers import Trainer, TrainingArguments
from transformers import DataCollatorWithPadding
import numpy as np
from sklearn.metrics import accuracy_score, f1_score

# Data collator that will pad sequences to the maximum input length
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

# Define compute_metrics function for evaluation during training
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    accuracy = accuracy_score(labels, predictions)
    f1 = f1_score(labels, predictions, average='weighted') # 'weighted' is good for imbalanced classes
    return {"accuracy": accuracy, "f1_score": f1}

# Initialize the Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train_dataset,
    eval_dataset=tokenized_test_dataset,
    tokenizer=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics,
)

print("Trainer initialized successfully with model, training arguments, datasets, tokenizer, data collator, and compute_metrics.")
```

**Reasoning**:
The previous `Trainer` initialization failed because the `Trainer` class no longer accepts a `tokenizer` argument directly in recent versions of the Hugging Face Transformers library. The tokenizer should be passed to the `DataCollatorWithPadding` as it is already being done. I will remove the `tokenizer` argument from the `Trainer` constructor.



```python
from transformers import Trainer, TrainingArguments
from transformers import DataCollatorWithPadding
import numpy as np
from sklearn.metrics import accuracy_score, f1_score

# Data collator that will pad sequences to the maximum input length
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

# Define compute_metrics function for evaluation during training
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    accuracy = accuracy_score(labels, predictions)
    f1 = f1_score(labels, predictions, average='weighted') # 'weighted' is good for imbalanced classes
    return {"accuracy": accuracy, "f1_score": f1}

# Initialize the Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train_dataset,
    eval_dataset=tokenized_test_dataset,
    data_collator=data_collator,
    compute_metrics=compute_metrics,
)

print("Trainer initialized successfully with model, training arguments, datasets, data collator, and compute_metrics.")
```

**Reasoning**:
The Trainer has been successfully initialized, and the next step is to initiate the training process using the `train()` method. This will start the fine-tuning of the DistilBERT model on the prepared dataset.



```python
trainer.train()

print("Training complete.")
```
